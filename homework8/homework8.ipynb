```python
import pandas as pd
```

```python
data = pd.read_csv("../course_assignments/homework8/state_data.csv")
```

```python
#4: print shape (rows, columns)
print(data.shape)

#5: print first and last 5 rows
print(data.head())
print(data.tail())
#Answers to questions:
#columns represent state abbr, latitude, longitude, and state's full name
#first state: alaska
#last state: wyoming
#has district of columbia and puerto rico
```

```python
#6: median latitude and logitude
median_lat = data["latitude"].median()
median_lon = data["longitude"].median()
print("median latitude is ", median_lat)
print("median longitude is ", median_lon)

#7/8: abs value distance to median lat/lon for each state (new col)
data["lat_dist"] = (data["latitude"]-median_lat).abs()
data["lon_dist"] = (data["longitude"]-median_lon).abs()
```

```python
#9: print new shape:
print(data.shape)
print("rows: ", data.shape[0])
print("cols: ", data.shape[1])
```

```python
#10:
closest_lat_row = data.loc[data["lat_dist"].idxmin()]
farthest_lat_row = data.loc[data["lat_dist"].idxmax()]
closest_lon_row = data.loc[data["lon_dist"].idxmin()]
farthest_lon_row = data.loc[data["lon_dist"].idxmax()]

print(f"Closest to median latitude: {closest_lat_row['state']} by {closest_lat_row['lat_dist']:.2f}")
print(f"Farthest from median latitude: {farthest_lat_row['state']} by {farthest_lat_row['lat_dist']:.2f}")
print(f"Closest to median longitude: {closest_lon_row['state']} by {closest_lon_row['lon_dist']:.2f}")
print(f"Farthest from median longitude: {farthest_lon_row['state']} by {farthest_lon_row['lon_dist']:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("attention")
df
```

```python
#5: drop columns with NaN values:
df = df.dropna(axis=1)
df.info()
```

```python
#6: analysis
print("shape: ", df.shape)
print("num of rows: ", df.shape[0])
print("num of columns: ", df.shape[1])

print("first five rows: ", df.head())
print("last five rows: ", df.tail())
print("column names: ", df.columns.tolist())
```

```python
# 7 and 8
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# top left: scatterplot 1
axes[0, 0].scatter(df["subject"], df["score"], color="blue", label="Score by subject")
axes[0, 0].set_title("Subject vs Score")
axes[0, 0].set_xlabel("Subject ID")
axes[0, 0].set_ylabel("Score (points)")
axes[0, 0].legend()

# top right: scatterplot 2
# using solutions and score as a different pair
axes[0, 1].scatter(df["solutions"], df["score"], color="red")
axes[0, 1].set_title("Solutions vs Score")
axes[0, 1].set_xlabel("Number of solutions")
axes[0, 1].set_ylabel("Score (points)")

# bottom left: histogram
sns.histplot(data=df, x="score", bins=10, color="green", ax=axes[1, 0])
axes[1, 0].set_title("Distribution of Scores")
axes[1, 0].set_xlabel("Score (points)")
axes[1, 0].set_ylabel("Count")

# bottom right: categorical plot (boxplot)
sns.boxplot(data=df, x="attention", y="score", palette="pastel", ax=axes[1, 1])
axes[1, 1].set_title("Score by Attention Type")
axes[1, 1].set_xlabel("Attention Type")
axes[1, 1].set_ylabel("Score (points)")

plt.tight_layout()
plt.show()
```

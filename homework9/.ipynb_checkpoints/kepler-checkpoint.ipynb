```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv("solar_system.csv", index_col=0)

print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("First five rows:")
print(df.head())

print("Last five rows:")
print(df.tail())

print("Column names:")
print(df.columns)

print("Row names:")
print(df.index)
```

```python
#5 extract the rows as variables
orbital_period = df.iloc[0].astype(float)
perihelion = df.iloc[1].astype(float)
aphelion = df.iloc[2].astype(float)

print("Orbital period:")
print(orbital_period)

print("Perihelion:")
print(perihelion)

print("Aphelion:")
print(aphelion)
```

```python
#defining stuff:
#perihelion: point in planet's orbit where it's closest to the sun
#aphelion: point in a planet's orbit where it's farthest from sun
#semi-major axis: half of the longest diameter of elliptical orbit (avg of above 2)
```

```python
semi_major_axis = (perihelion + aphelion) / 2
print(semi_major_axis)

semi_major_axis_row = pd.Series(semi_major_axis, name="Semi-major axis")
df = pd.concat([df, semi_major_axis_row.to_frame().T])

print(df)
```

```python
print("New shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

plt.figure(figsize=(8, 5))
plt.scatter(semi_major_axis, orbital_period, label="Planets")

plt.xlabel("Semi-major axis (10^6 km)")
plt.ylabel("Orbital period (days)")
plt.title("Semi-major axis vs Orbital period")
plt.legend()
plt.show()
```

```python
def keplers_third_law(a, m):
    return m * a**1.5
popt, pcov = curve_fit(keplers_third_law, semi_major_axis, orbital_period)

m_fit = popt[0]
print("Best-fit m:", m_fit)

print(f"Fitted model: T = {m_fit:.2f} * a^1.5")

residuals = orbital_period - keplers_third_law(semi_major_axis, m_fit)
chi2 = np.sum(residuals**2)
dof = len(orbital_period) - 1
chi2_reduced = chi2 / dof

print(f"Reduced chi-squared: {chi2_reduced:.2f}")

a_smooth = np.linspace(semi_major_axis.min(), semi_major_axis.max(), 200)
T_smooth = keplers_third_law(a_smooth, m_fit)

plt.figure(figsize=(8, 5))
plt.scatter(semi_major_axis, orbital_period, label="Planet data")
plt.plot(a_smooth, T_smooth, label="Fitted model")

plt.xlabel("Semi-major axis (10^6 km)")
plt.ylabel("Orbital period (days)")
plt.title(f"Kepler Fit, Reduced chi-squared = {chi2_reduced:.2f}")
plt.legend()
plt.show()

#save!
df.to_csv("kepler.csv")
```

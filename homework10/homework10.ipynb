```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from astropy import units as u
from astropy import constants as const
```

```python
df = pd.read_csv("solar_system.csv")
df
```

```python
print("Before transposing:", df.shape)
```

```python
df = df.set_index("Attribute").T
#Makes the attribute column become the row labels, then transposes the table.
#This makes each row a planet instead of each row being an attribute
print("After set_index and transpose:", df.shape)

df.index.name = "Planet"
#Names the row index planet
print("After naming index:", df.shape)

df.reset_index(inplace=True)
#turns the planet names from the index back into a normal column
print("After reset_index:", df.shape)

df.columns.name = None
#removes the label/name above the column headers
print("After removing columns name:", df.shape)

#Why isn't the new shape (11, 20)?
#after reset_index(), the planet names become an extra column called "Planet."
#instead of having only the original attribute columns, you also have a planet-name column

df
```

```python
#print new column names:
print(df.columns.tolist())
```

```python
with_units = []
without_units = []

for col in df.columns:
    if "(" in col and ")" in col:
        with_units.append(col)
    else:
        without_units.append(col)

print("Columns with units:", len(with_units))
print(with_units)

print()

print("Columns without units:", len(without_units))
print(without_units)
```

```python
#When I used df.dtypes, every column is an object, meaning pandas is not storing the clumns as true numeric values yet. 
#Then when I looped through each column and found that the actual values inside every column are strings. 
#Although many of the values look like numbers they're actually stored as text

#(a): the actual values are strings. Every column contains 10 values with type <class 'str'>.

#(b): Storing numeric values as strings is a problem because pandas treas them as text instead of numbers. That means calculations may not work correctly. 
```

```python
#2.1.3: convert strings to numbers:
skip_cols = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in skip_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df.dtypes
```

```python
#3 types of data: object, float64, int64
    #object: text/string-like values
    #float64: decimal numbers
    #int64: integers
#some cols are still objects. For example, planet is still a object because it contains planet names. 
```

```python
#2.2
def attach_units(df, old_col, unit, new_col):
    new_values = []

    for val in df[old_col]:
        new_values.append(val * unit)

    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)

attach_units(df, "Mass (10^24kg)", 1e24 * u.kg, "Mass (kg)")
df["Mass (kg)"]
```

```python
#2.3: data analysis
semi_major_axis = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2
semi_major_axis = semi_major_axis * 1e6 * u.km
aphelion_index = df.columns.get_loc("Aphelion (10^6 km)")
df.insert(aphelion_index + 1, "Semi-Major Axis (km)", semi_major_axis)

print(df.columns.tolist())
df["Semi-Major Axis (km)"]
```

```python
#convert orbital period to years

#check the exact name
df.columns.tolist()
```

```python
orbital_years = []

for val in df["Orbital Period (days)"]:
    orbital_years.append((val * u.day).to(u.year))

df["Orbital Period (years)"] = orbital_years
df.drop(columns=["Orbital Period (days)"], inplace=True)
```

```python
print(df.columns.tolist())
df["Orbital Period (years)"]
```

```python
favorite = "Earth"

period = df.loc[df["Planet"] == favorite, "Orbital Period (years)"].iloc[0]
print(f"{favorite}'s orbital period is {period.value:.4f} years")
```

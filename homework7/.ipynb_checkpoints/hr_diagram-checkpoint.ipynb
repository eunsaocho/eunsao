```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
#3.2.1
print("Shape: ", data.shape)
print("Rows: ", data.shape[0])
```

```python
#3.2.3
magnitude = data[:, 0]
temperature = data[:, 1]

#3.2.4 print min and max of magnitude
min_mag = np.min(magnitude)
max_mag = np.max(magnitude)
min_mag_i = np.argmin(magnitude)
max_mag_i = np.argmax(magnitude)
print("Min magnitude: ", min_mag, "at row ", min_mag_i)
print("Max magnitude: ", max_mag, "at row ", max_mag_i)

#3.2.5 min max temp
min_temp = np.min(temperature)
max_temp = np.max(temperature)
min_temp_i = np.argmin(temperature)
max_temp_i = np.argmax(temperature)
print("Min temp: ", min_temp, "at row ", min_temp_i)
print("Max temp: ", max_temp, "at row ", max_temp_i)

#3.2.6 brightest and dimmest star
print("Brightest star is at row ", min_mag_i, " with magnitude ", magnitude[min_mag_i])
print("Dimmest star is at row ", max_mag_i, " with magnitude ", magnitude[max_mag_i])
```

```python
#3.3
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))

#3.3.3: scatter all stars
plt.scatter(log_temperature, log_magnitude, label="Stars")

#3.3.4: invert axes
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

#3.3.5
plt.scatter(log_temperature[min_mag_i], log_magnitude[min_mag_i], color="black", marker="*", s=300, label="Brightest star")
plt.scatter(log_temperature[max_mag_i], log_magnitude[max_mag_i], color="purple", marker="*", s=300, label="Dimmest star")
plt.scatter(log_temperature[max_temp_i], log_magnitude[max_temp_i], color="green", marker="*", s=300, label="Hottest star")
plt.scatter(log_temperature[min_temp_i], log_magnitude[min_temp_i], color="orange", marker="*", s=300, label="Coldest star")

plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show()
```
